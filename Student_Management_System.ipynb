```python
"""
STUDENT MANAGEMENT SYSTEM
A complete Python application for managing student records.
"""

students = {}
next_id = 1


def add_student():
    """Add a new student to the database."""
    global next_id

    print("\n" + "-" * 40)
    print("ADD NEW STUDENT")
    print("-" * 40)

    name = input("Enter student name: ").title()
    age = int(input("Enter age: "))
    course = input("Enter course: ").upper()

    while True:
        grade = float(input("Enter grade (0-100): "))
        if 0 <= grade <= 100:
            break
        print("Grade must be between 0 and 100!")

    if grade >= 90:
        letter_grade = "A+"
    elif grade >= 80:
        letter_grade = "A"
    elif grade >= 70:
        letter_grade = "B"
    elif grade >= 60:
        letter_grade = "C"
    else:
        letter_grade = "F"

    students[next_id] = {
        "name": name,
        "age": age,
        "course": course,
        "grade": grade,
        "letter_grade": letter_grade
    }

    print(f"\nStudent added! ID: {next_id}")
    next_id += 1
    input("\nPress Enter to continue...")


def view_all_students():
    """Display all students in the database."""
    print("\n" + "-" * 50)
    print("ALL STUDENTS")
    print("-" * 50)

    if len(students) == 0:
        print("No students found.")
    else:
        print(f"\n{'ID':<5} {'Name':<15} {'Age':<5} {'Course':<12} {'Grade':<8} {'Letter'}")
        print("-" * 55)

        for student_id, info in students.items():
            print(f"{student_id:<5} {info['name']:<15} {info['age']:<5} {info['course']:<12} {info['grade']:<8} {info['letter_grade']}")

        print("-" * 55)
        print(f"Total students: {len(students)}")

    input("\nPress Enter to continue...")


def search_student():
    """Search for a student by ID or name."""
    print("\n" + "-" * 40)
    print("SEARCH STUDENT")
    print("-" * 40)

    print("Search by:")
    print("  1. Student ID")
    print("  2. Name")

    choice = input("\nEnter choice (1/2): ")

    if choice == "1":
        try:
            student_id = int(input("Enter student ID: "))
            if student_id in students:
                info = students[student_id]
                print("\nSTUDENT FOUND:")
                print(f"  ID: {student_id}")
                print(f"  Name: {info['name']}")
                print(f"  Age: {info['age']}")
                print(f"  Course: {info['course']}")
                print(f"  Grade: {info['grade']} ({info['letter_grade']})")
            else:
                print(f"No student found with ID: {student_id}")
        except:
            print("Invalid ID!")

    elif choice == "2":
        search_name = input("Enter student name: ").title()
        found = False

        for student_id, info in students.items():
            if info['name'] == search_name:
                print(f"  ID: {student_id} | Grade: {info['grade']} ({info['letter_grade']})")
                found = True

        if not found:
            print(f"No student found with name: {search_name}")

    else:
        print("Invalid choice!")

    input("\nPress Enter to continue...")


def update_student():
    """Update a student's grade."""
    print("\n" + "-" * 40)
    print("UPDATE STUDENT GRADE")
    print("-" * 40)

    try:
        student_id = int(input("Enter student ID to update: "))

        if student_id in students:
            info = students[student_id]
            print(f"\nCurrent student: {info['name']}")
            print(f"Current grade: {info['grade']} ({info['letter_grade']})")

            while True:
                new_grade = float(input("\nEnter new grade (0-100): "))
                if 0 <= new_grade <= 100:
                    break
                print("Grade must be between 0 and 100!")

            students[student_id]['grade'] = new_grade

            if new_grade >= 90:
                students[student_id]['letter_grade'] = "A+"
            elif new_grade >= 80:
                students[student_id]['letter_grade'] = "A"
            elif new_grade >= 70:
                students[student_id]['letter_grade'] = "B"
            elif new_grade >= 60:
                students[student_id]['letter_grade'] = "C"
            else:
                students[student_id]['letter_grade'] = "F"

            print(f"\nGrade updated! New grade: {new_grade} ({students[student_id]['letter_grade']})")
        else:
            print(f"No student found with ID: {student_id}")
    except:
        print("Invalid input!")

    input("\nPress Enter to continue...")


def delete_student():
    """Remove a student from the database."""
    print("\n" + "-" * 40)
    print("DELETE STUDENT")
    print("-" * 40)

    try:
        student_id = int(input("Enter student ID to delete: "))

        if student_id in students:
            info = students[student_id]
            print(f"\nWARNING: You are about to delete:")
            print(f"   Name: {info['name']}")
            print(f"   ID: {student_id}")

            confirm = input("\nAre you sure? (yes/no): ").lower()

            if confirm == "yes":
                del students[student_id]
                print(f"\nStudent {info['name']} has been deleted!")
            else:
                print("Deletion cancelled.")
        else:
            print(f"No student found with ID: {student_id}")
    except:
        print("Invalid input!")

    input("\nPress Enter to continue...")


def show_statistics():
    """Calculate and display class statistics."""
    print("\n" + "-" * 50)
    print("CLASS STATISTICS")
    print("-" * 50)

    if len(students) == 0:
        print("No students to analyze!")
    else:
        grades = [info['grade'] for info in students.values()]

        average = sum(grades) / len(grades)
        highest = max(grades)
        lowest = min(grades)

        a_count = sum(1 for g in grades if g >= 80)
        b_count = sum(1 for g in grades if 70 <= g < 80)
        c_count = sum(1 for g in grades if 60 <= g < 70)
        f_count = sum(1 for g in grades if g < 60)

        print(f"\nTotal Students: {len(students)}")
        print(f"Average Grade: {average:.2f}")
        print(f"Highest Grade: {highest}")
        print(f"Lowest Grade: {lowest}")

        print(f"\nGRADE DISTRIBUTION:")
        print(f"  A Grade (80+): {a_count} student(s)")
        print(f"  B Grade (70-79): {b_count} student(s)")
        print(f"  C Grade (60-69): {c_count} student(s)")
        print(f"  F Grade (<60): {f_count} student(s)")

        top_student = max(students.items(), key=lambda x: x[1]['grade'])
        print(f"\nTOP PERFORMER:")
        print(f"  {top_student[1]['name']} (ID: {top_student[0]})")
        print(f"  Grade: {top_student[1]['grade']} ({top_student[1]['letter_grade']})")

    input("\nPress Enter to continue...")


def save_to_file():
    """Save student data to a text file."""
    if len(students) == 0:
        print("No data to save!")
    else:
        with open("students_data.txt", "w") as file:
            file.write("=" * 60 + "\n")
            file.write("STUDENT MANAGEMENT SYSTEM - EXPORTED DATA\n")
            file.write("=" * 60 + "\n\n")

            for student_id, info in students.items():
                file.write(f"ID: {student_id}\n")
                file.write(f"Name: {info['name']}\n")
                file.write(f"Age: {info['age']}\n")
                file.write(f"Course: {info['course']}\n")
                file.write(f"Grade: {info['grade']} ({info['letter_grade']})\n")
                file.write("-" * 30 + "\n")

            file.write(f"\nTotal Students: {len(students)}")

        print(f"\nData saved to 'students_data.txt'")

    input("\nPress Enter to continue...")


def main():
    """Main application menu."""
    running = True

    while running:
        print("\n" + "=" * 55)
        print("STUDENT MANAGEMENT SYSTEM")
        print("=" * 55)
        print(f"Active Students: {len(students)}")
        print("-" * 55)
        print("  1. Add Student")
        print("  2. View All Students")
        print("  3. Search Student")
        print("  4. Update Student Grade")
        print("  5. Delete Student")
        print("  6. Class Statistics")
        print("  7. Save & Export Data")
        print("  8. Exit")
        print("-" * 55)

        choice = input("Enter your choice (1-8): ")

        if choice == "1":
            add_student()
        elif choice == "2":
            view_all_students()
        elif choice == "3":
            search_student()
        elif choice == "4":
            update_student()
        elif choice == "5":
            delete_student()
        elif choice == "6":
            show_statistics()
        elif choice == "7":
            save_to_file()
        elif choice == "8":
            print("\nThank you for using Student Management System!")
            print(f"Total students managed: {len(students)}")
            running = False
        else:
            print("Invalid choice! Please enter 1-8")
            input("Press Enter to continue...")


if __name__ == "__main__":
    main()
```
